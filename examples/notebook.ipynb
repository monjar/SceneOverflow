# SceneOverflow in a notebook
A clip renders itself as a player + thumbnails + timeline.

```python
from sceneoverflow import Project
p = Project('media')
talk = p.videos['talk']
talk.head('6s')
```

```python
cut = talk.remove(('2s', '4s')).with_audio(p.sounds[0], gain=0.3)
print(cut.describe())
cut
```
